```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_breast_cancer
```

```python
data=load_breast_cancer()
#Convert dataset into a Pandas DataFrame
df = pd.DataFrame(data.data, columns=data.feature_names)
# Add target column
df['target'] = data.target

df.head()
df.columns
```

```python
#Check dataset shape
print(df.shape)

# Check column information
#print(df.info())

# Check statistical summary
#print(df.describe())

# Check target distribution
print(df['target'].value_counts())
print("Target", data.target_names)
```

Observations

- 357 benign cases.

```python
#check for null value
df.isnull().sum()
#check the information about the table
df.info()
#take the statistical information
df.describe()
#drop duplicates if any
df.drop_duplicates(inplace=True)
df.shape
```

```python
from sklearn.model_selection import train_test_split, cross_val_score, KFold
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
```

```python
x=data.data
y=data.target
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)

model= SVC(kernel="rbf")
model.fit(x_train,y_train)

y_pred=model.predict(x_test)
y_pred

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import RandomizedSearchCV


from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
# KNN
knn = KNeighborsClassifier(n_neighbors=5)
params={
    'n_neighbors':[3,5,7,9,11],
    'weights':['uniform','distance'],
    'metric':['euclidean','manhattan']
}
grid=GridSearchCV(
    estimator=knn,
    param_grid=params,
    cv=5,
    scoring='accuracy'
)

grid.fit(x_train,y_train)
```

```python
print(grid.best_params_)
print(grid.best_score_)
```

KNN have best score of 0.9648351648351647 when the params are {'metric': 'manhattan', 'n_neighbors': 5, 'weights': 'uniform'}

```python
model_params = {
    'svm': {'model': SVC(gamma='auto'), 'params': {'C': [1, 10, 20], 'kernel': ['rbf', 'linear']}},
    'random_forest': {'model': RandomForestClassifier(random_state=42), 'params': {'n_estimators': [100, 200],'max_depth': [5, 10, None],'min_samples_split': [2, 5]}},
    'logistic_regression': {'model': LogisticRegression(max_iter=1000), 'params': {'C': [1, 5, 10]}},
    'kneighbor': {'model': KNeighborsClassifier(), 'params': {'n_neighbors': [1,3, 5, 7, 9, 11],'weights':['uniform','distance'],'metric':['euclidean','manhattan']}},
    'decision_tree': {'model': DecisionTreeClassifier(random_state=42), 'params': {'max_depth': [3, 5, 7, 10, None],'min_samples_split': [2, 5, 10],'criterion': ['gini', 'entropy']}}
}
```

```python
scores = []
for model_name, mp in model_params.items():
    clf = GridSearchCV(mp['model'], mp['params'], cv=5, return_train_score=False)
    clf.fit(x_train, y_train)
    scores.append({
        'model': model_name,
        'best_score': clf.best_score_,
        'test_score': clf.score(x_test, y_test),
        'best_params': clf.best_params_
    })

df = pd.DataFrame(scores)
df.sort_values('best_score', ascending=False)
```

Logistic Regression

It suits this dataset because the two classes are almost linearly separable, and it is fast and interpretable.

Decision Tree

A Decision Tree asks a series of yes/no questions about the features, splitting the data until each group is mostly one class. It is easy to explain and needs no scaling, but it overfits easily, which is why it performed worst here.

Random Forest

A Random Forest trains many decision trees on random samples of the data and takes a majority vote. It suits this dataset because it handles the many correlated features well and reduces the overfitting of a single tree.

Support Vector Machine (SVM)

SVM finds the boundary that separates the two classes with the widest margin, using kernels for curved boundaries. It works well on small datasets with many numeric features, as long as they are scaled.

k-Nearest Neighbors (k-NN)

k-NN classifies a sample by the majority class among its k closest training samples. It suits this dataset because similar tumours tend to share a diagnosis, but it needs scaled features since it relies on distance.

Best: Logistic Regression. It has the highest CV score (0.978) and ties SVM for the highest test score (0.9737).

Worst: Decision Tree. It has the lowest CV score (0.947) and the lowest test score (0.956).

SVM is a close second. It matches Logistic Regression on the test score, and its CV score is only 0.004 lower, so the gap between the top two is small.
