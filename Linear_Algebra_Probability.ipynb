## Linear Algebra and Probability

1. Load the dataset and handle any missing data by replacing null values with the mean
value of the respective column. (Score 2)

```python
import pandas as pd
import numpy as np


# Load the dataset
df = pd.read_csv("/content/winequality-red.csv")
# Display the first five rows
print(df.head())

# Check the shape of the dataset
print("Dataset shape:", df.shape)

# Check missing values
print(df.isnull().sum())

# Replace missing values with the mean of each column
df = df.fillna(df.mean(numeric_only=True))

# Verify missing values after replacement
print(df.isnull().sum())
```

2. Extract the following columns as vectors: alcohol,citric acid. (Score 2)

```python

# Extract vectors
alcohol = df["alcohol"].to_numpy()
citric_acid = df["citric acid"].to_numpy()

print("Alcohol vector:")
print(alcohol)

print("\nCitric acid vector:")
print(citric_acid)

print("Alcohol shape:", alcohol.shape)
print("Citric acid shape:", citric_acid.shape)
```

3. Select two features (e.g., alcohol and density) from the dataset and calculate the
covariance matrix using np.cov(X.T), where X is the feature matrix consisting of the
selected columns.(Score 2)

```python
# Select two features
X = df[["alcohol", "density"]].values

# Calculate covariance matrix
cov_matrix = np.cov(X.T)

print("Covariance Matrix:")
print(cov_matrix)
```

4. Perform eigen decomposition on the covariance matrix you computed in question 3.
Identify and interpret the results:Identify the top 2 eigenvalues of the covariance
matrix,Identify the corresponding eigenvectors.(Score 2)


```python
# Perform eigen decomposition
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

# Sort eigenvalues in descending order
idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

# Display results
print("Covariance Matrix:")
print(cov_matrix)

print("\nTop 2 Eigenvalues:")
print(eigenvalues[:2])

print("\nCorresponding Eigenvectors:")
print(eigenvectors[:, :2])
```

5. Which wine quality is most common in the dataset? How can you interpret the
distribution of wine quality scores? (Score 2)

```python
# Count the occurrences of each wine quality score
quality_counts = df["quality"].value_counts()

print("Wine Quality Distribution:")
print(quality_counts)

# Identify the most common quality score
most_common_quality = quality_counts.idxmax()
print("\nMost common wine quality:", most_common_quality)
print("Number of wines:", quality_counts.max())
```
