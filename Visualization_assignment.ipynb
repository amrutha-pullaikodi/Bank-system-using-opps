## Visualization with Seaborn and Matplotlib (Assignment)


Use the “Titanic” dataset from Seaborn and do the following:
1. Countplot: Passenger Count by Embarked Port
● Create a countplot to display the number of passengers who boarded from each
port (embarked), separated by sex.

```python
#importing the libraries
import seaborn as sns
sns.set(style="darkgrid")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import warnings
warnings.filterwarnings("ignore")
```

```python
titanic=sns.load_dataset("titanic")
#titanic=titanic.dropna()
```

```python
sns.countplot(data=titanic,x="embarked",hue="sex")
plt.title("Passenger Count by Embarked Port - Separated by Sex")
plt.xlabel("Embarked Port")
plt.ylabel("Number of Passengers")
plt.show()
print("Male passengers outnumber female passengers at all three boarding ports. S has the highest number of passengers, followed by C and Q.")
```

2. Swarmplot: Age Distribution across Classes
● Create a swarmplot to visualize the distribution of age across passenger classes
(pclass), categorized by sex.

```python
sns.swarmplot(data=titanic,y="age",x="pclass",hue="sex")
plt.title("Distribution of age across Passenger Classes")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.show()
print("Passengers of different ages are present in all three classes. First-class passengers tend to be older on average, while the third class includes more younger passengers and children.")
```

3. Pie Chart: Gender Distribution
● Create a pie chart using Matplotlib to show the percentage of male vs. female
passengers. - Add percentage labels using autopct - Use explode for one slice to
highlight it.

```python
gender_count=titanic["sex"].value_counts()
print(gender_count)
plt.figure(figsize=(8,6))
plt.pie(gender_count,labels=gender_count.index,autopct="%.2f%%",explode=[0,0.1],startangle=140,shadow=True)
plt.title("Gender Distribution")
plt.axis("equal")
plt.show()
print("The database contain more male passengers(64%) than female passengers(35.24%).")
```

4. Heatmap: Correlation Matrix
● Create a heatmap to show correlations between age, fare, pclass, sibsp, parch,
and survived. (Use seaborn.heatmap with annotations)

```python
data=titanic[["age","fare","pclass","sibsp","parch","survived"]]
correlation_matrix=data.corr()
plt.figure(figsize=(8,6))
sns.heatmap(correlation_matrix,annot=True,cmap='coolwarm')
plt.title("Heatmap: Correlation Matrix")
plt.show()
print("Passenger class has a negative correlation with fare. Survival has a positive correlation with fare and a negative correlation with passenger class.")
```

5. Violin Plot: Fare by Class and Gender
● Create a violin plot to visualize fare distribution across classes, separated by
gender.

```python
plt.figure(figsize=(8,6))
sns.violinplot(titanic, x="pclass", y="fare", hue="sex")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.title("Violin Plot: Fare by Class and Gender")
plt.show()
print("Insights:First-class passengers generally paid higher fares than second- and third-class passengers. Fare distributions vary across passenger classes and gender, with some unusually high fares in first class.")
```

6. Scatter Plot: Age vs Fare
● Create a scatter plot to explore the relationship between age and fare, colored by
survival status.

● Use a seaborn scatterplot with hue='survived'. Each plot must have titles, labels,
and legends where needed.

```python
sns.relplot(x="age",y="fare",data=titanic,kind="scatter",hue="survived")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
print("Passengers who survived and those who did not are distributed across different ages and fares. Higher fares appear among some survivors, but age and fare alone do not perfectly predict survival.")
```
