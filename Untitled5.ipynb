```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#Load the dataset
abc=pd.read_csv("/content/ABC Company.xlsx - Sheet1.csv")
#abc
```

Preprocessing:
● Correct the data in the "height" column by replacing it with random
numbers between 150 and 180. Ensure data consistency and integrity
before proceeding with analysis. (1 mark)

```python
abc["Height"]=np.random.randint(150,181)
print(abc.shape)
abc.isnull().sum()
abc.duplicated().sum()
```

● Determine the distribution of employees across each team and calculate
the percentage split relative to the total number of employees. (2 marks)

```python
team_count=abc["Team"].value_counts()
#print(team_count)
team_perc=abc["Team"].value_counts(normalize=True)*100
#team_perc

team_analysis = pd.DataFrame({
    "Employee Count": team_count,
    "Percentage": team_perc.round(2)
})

display(team_analysis)
team_count.plot(kind="bar", color="blue")
plt.title("Employee Distribution Across Teams")
plt.xlabel("Team")
plt.ylabel("Number of Employees")
plt.show()
```

Segregate employees based on their positions within the company. (2
marks)

```python
position_count = abc["Position"].value_counts()

print(position_count)
sns.countplot(data=abc, x="Position", order=position_count.index)
plt.title("Employee Distribution by Position")
plt.xlabel("Position")
plt.ylabel("Number of Employees")
plt.show()
```

Identify the predominant age group among employees. (2 marks)

```python

plt.xlabel("Age group")
plt.ylabel("Frequency")
plt.xticks(range(19,43,2))
plt.hist(abc["Age"],bins=30);
print("Predominant age group:23-26")

```

● Discover which team and position have the highest salary expenditure. (2
marks)

```python

team_salary = abc.groupby("Team")["Salary"].sum().sort_values(
    ascending=False
)

position_salary = abc.groupby("Position")["Salary"].sum().sort_values(
    ascending=False
)
print(team_salary)
print(position_salary)
print("Team with highest salary expenditure:")
print(team_salary.idxmax(), team_salary.max())

print("\nPosition with highest salary expenditure:")
print(position_salary.idxmax(), position_salary.max())

#Visualization (Sub plot)
fig,ax=plt.subplots(nrows=1,ncols=2,figsize=(12,3))
ax[0].plot(position_salary)
ax[0].set_title("Total salary expenditure by position")
ax[0].set_ylabel("Salary (millions $)")
ax[0].set_xlabel("Position")
ax[0].tick_params(axis="x", rotation=90)

ax[1].plot(team_salary)
ax[1].set_title("Total salary expenditure by team")
ax[1].set_ylabel("Salary (millions $)")
ax[1].set_xlabel("Team")
ax[1].tick_params(axis="x", rotation=90)


```

 Investigate if there's any correlation between age and salary, and
represent it visually. (2 marks)

```python
df1=abc[["Age","Salary"]]
Correlation_mtx=df1.corr()
Correlation_mtx
plt.title("Correlation between age and salary")
sns.heatmap(Correlation_mtx,annot=True,cmap='coolwarm')
plt.show()
print("A positive correlation indicates that older employees tend to have higher salaries; a negative correlation indicates the opposite. A value close to zero indicates little linear relationship.")
```

A positive correlation indicates that older employees tend to have higher salaries; a negative correlation indicates the opposite. A value close to zero indicates little linear relationship.
